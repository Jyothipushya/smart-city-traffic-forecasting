```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv('/content/smart-city-traffic-patterns/smart-city-traffic-patterns/train_aWnotuB.csv')

# Convert DateTime
df['DateTime'] = pd.to_datetime(df['DateTime'])

# Extract features
df['Hour'] = df['DateTime'].dt.hour
df['DayOfWeek'] = df['DateTime'].dt.dayofweek
df['Month'] = df['DateTime'].dt.month
df['Year'] = df['DateTime'].dt.year
df['IsWeekend'] = df['DayOfWeek'].apply(lambda x: 1 if x >= 5 else 0)

print("Done!")
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Graph 1 - Traffic by Hour
plt.figure(figsize=(12,5))
df.groupby('Hour')['Vehicles'].mean().plot(kind='bar', color='steelblue')
plt.title('Average Traffic by Hour of Day')
plt.xlabel('Hour')
plt.ylabel('Average Vehicles')
plt.tight_layout()
plt.savefig('/content/traffic_by_hour.png')
plt.show()

# Graph 2 - Traffic by Junction
plt.figure(figsize=(8,5))
df.groupby('Junction')['Vehicles'].mean().plot(kind='bar', color='coral')
plt.title('Average Traffic per Junction')
plt.xlabel('Junction')
plt.ylabel('Average Vehicles')
plt.tight_layout()
plt.savefig('/content/traffic_by_junction.png')
plt.show()

# Graph 3 - Weekday vs Weekend
plt.figure(figsize=(6,5))
df.groupby('IsWeekend')['Vehicles'].mean().plot(kind='bar', color=['green','orange'])
plt.title('Weekday vs Weekend Traffic')
plt.xticks([0,1], ['Weekday','Weekend'], rotation=0)
plt.ylabel('Average Vehicles')
plt.tight_layout()
plt.savefig('/content/weekday_vs_weekend.png')
plt.show()

print("All graphs saved!")
```
